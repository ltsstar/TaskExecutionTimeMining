```python
import pm4py
import sys
sys.path.append('../../../TaskExecutionTimeMining/')
from event_log_transformer import *
import pandas
pandas.set_option('display.max_columns', None)
pandas.set_option('display.max_rows', 200)
```

```python
file_path = '../../../../data/BPI_Challenge_2019.xes'
event_log = pm4py.read_xes(file_path)
event_log['id'] = event_log.index
#event_log['Complete Timestamp'] = pandas.to_datetime(event_log['Complete Timestamp'])
```

```python
event_log
```

```python
start_end_event_log = TransformEventLog.start_end_event_log_next(event_log,
                                 merge_activity_on = ['case:concept:name'],
                                 timestamp_name = 'time:timestamp',
                                 start_name_gen = '_start',
                                 complete_name_gen = '_complete',
                                 unique_column = 'id')
```

```python
# Seconds in day
start_end_event_log = TransformEventLog.seconds_in_day(start_end_event_log, 'time:timestamp_start')
start_end_event_log = TransformEventLog.day_of_week(start_end_event_log, 'time:timestamp_start')
```

```python
# Resource counts
start_end_event_log = TransformEventLog.value_count_per_case_without_lifecycle(start_end_event_log,
                                                             column_name = 'concept:name_start',
                                                            case_name = 'case:concept:name',
                                                            timestamp_name = 'time:timestamp_start',
                                                            unique_id = 'id_start')
start_end_event_log = TransformEventLog.value_count_per_case_without_lifecycle(start_end_event_log,
                                                             column_name = 'org:resource_start',
                                                            case_name = 'case:concept:name',
                                                            timestamp_name = 'time:timestamp_start',
                                                            unique_id = 'id_start')
start_end_event_log = start_end_event_log.sort_values(by='time:timestamp_start')
```

```python
start_end_event_log
```

```python
# inter-case information
inter_case_event_log = TransformEventLog.inter_instance_encoding_last_events(start_end_event_log,
                                case_name = 'case:concept:name',
                                timestamp_name = 'time:timestamp_start',
                                concept_name = 'concept:name_start',
                                state_prefix = 'intercase_n_1__',
                                n=1)

inter_case_event_log = TransformEventLog.inter_instance_encoding_last_events(inter_case_event_log,
                                case_name = 'case:concept:name',
                                timestamp_name = 'time:timestamp_start',
                                concept_name = 'concept:name_start',
                                state_prefix = 'intercase_n_3__',
                                n=3)
```

```python
import random
random.seed(42)

train, test = pm4py.ml.split_train_test(inter_case_event_log, case_id_key='case:concept:name')
```

```python
print(train.shape, test.shape)
print(train['case:concept:name'].unique().shape, test['case:concept:name'].unique().shape)
```

```python
train.to_csv('../../transformed_event_logs/BPIC_19_train.csv', index=False, date_format='%Y-%m-%d %H:%M:%S.%f')
train.to_pickle('../../transformed_event_logs/BPIC_19_train.pickle')
```

```python
test.to_csv('../../transformed_event_logs/BPIC_19_test.csv', index=False, date_format='%Y-%m-%d %H:%M:%S.%f')
test.to_pickle('../../transformed_event_logs/BPIC_19_test.pickle')
```
