# Prefix duration predictor demo

This notebook replicates the explicit list-of-dictionaries example shown in `prefix_duration_predictor.py` and runs the predictor end to end.

```python
from pathlib import Path
import sys

NOTEBOOK_DIR = Path.cwd().resolve()

def find_repo_root(marker: str = "TaskExecutionTimeMining") -> Path:
    for candidate in (NOTEBOOK_DIR,) + tuple(NOTEBOOK_DIR.parents):
        if candidate.name == marker:
            return candidate
    raise RuntimeError(f"Unable to locate repo root named {marker}")

REPO_ROOT = find_repo_root()
LOCAL_LIB = NOTEBOOK_DIR

for path in (REPO_ROOT, LOCAL_LIB):
    if str(path) not in sys.path:
        sys.path.insert(0, str(path))

from prefix_duration_predictor import PrefixDurationPredictor
```

```python
example_prefix = [
    {
        "Complete Timestamp_start": "2024-01-03 11:25:00.000",
        "Activity_start": "Register",
        "Resource_start": "clerk_7",
        "seconds_in_day": 10000,
        "day_in_week": 1,
        "duration_seconds": 0,
    },
    {
        "Complete Timestamp_start": "2024-01-03 12:05:00.000",
        "Activity_start": "Examine",
        "Resource_start": "clerk_4",
        "seconds_in_day": 43500,
        "day_in_week": 1,
    }
]
predictor = PrefixDurationPredictor(device='cpu')
prediction = predictor.predict(example_prefix)
print(prediction)
```
