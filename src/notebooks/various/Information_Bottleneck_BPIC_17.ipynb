```python
import sys
import pickle
sys.path.append('../../TaskExecutionTimeMining/')
from mutual_information import *
from information_bottleneck import *
from sklearn.metrics import mutual_info_score
from sklearn.feature_selection import mutual_info_regression

import pandas as pd
import numpy as np
#np.seterr(divide='ignore', invalid='ignore')
```

```python
with open("../transformed_event_logs/BPIC_2017_all_train.pickle", "rb") as f:
    event_log = pickle.load(f)

# numerical attributes : duration, seconds_in_day, day_in_week
numerical_attributes = [
    'duration_seconds',
    'seconds_in_day',
    'case:RequestedAmount_start'
]

transformed_event_log = event_log.copy()

for num_attr in numerical_attributes:
    transformed_event_log[num_attr] = np.log(transformed_event_log[num_attr]+1)
    transformed_event_log[num_attr] = (transformed_event_log[num_attr] - transformed_event_log[num_attr].mean()) / transformed_event_log[num_attr].std()
```

```python
cluster_assignments, cluster_ranges = information_bottleneck_clustering(transformed_event_log['day_of_week'], transformed_event_log['duration_seconds'].to_numpy(), n_clusters=16,
                                                                        x_is_discrete=True,
                                                                        n_bins_x=2048, n_bins_y=2048)
```

```python
cluster_assignments, cluster_ranges = information_bottleneck_clustering(transformed_event_log['concept:name'], transformed_event_log['duration_seconds'].to_numpy(), n_clusters=16,
                                                                        x_is_discrete=True,
                                                                        n_bins_x=2048, n_bins_y=2048)
```

```python
cluster_assignments, cluster_ranges = information_bottleneck_clustering(transformed_event_log['org:resource_start'], transformed_event_log['duration_seconds'].to_numpy(), n_clusters=16,
                                                                        x_is_discrete=True,
                                                                        n_bins_x=2048, n_bins_y=2048)
```

```python
cluster_assignments, cluster_ranges = information_bottleneck_clustering(transformed_event_log['seconds_in_day'].to_numpy(), transformed_event_log['duration_seconds'].to_numpy(), n_clusters=16,
                                                                        n_bins_x=512, n_bins_y=2048)
```

```python
cluster_assignments, cluster_ranges = information_bottleneck_clustering(transformed_event_log['seconds_in_day'].to_numpy(), transformed_event_log['duration_seconds'].to_numpy(), n_clusters=16,
                                                                        n_bins_x=128, n_bins_y=512)
```
