```python
import pickle
import re
from pathlib import Path

import sys
sys.path.append('../../../TaskExecutionTimeMining/')
from quantile_regression import QuantileRegression

# Process multiple pickle files
pkl_files = [
    #("quantile_regression_models.pkl", "qrm__"),
    ("quantile_regression_models.pkl", "qrm__")
]

for pkl_filename, prefix in pkl_files:
    pkl_path = Path(pkl_filename)
    
    if not pkl_path.exists():
        print(f"{pkl_path} not found, skipping")
        continue
    
    with pkl_path.open("rb") as f:
        models = pickle.load(f)
    
    if not isinstance(models, dict):
        raise TypeError(f"Loaded object from {pkl_filename} is not a dict of models")
    
    saved = []
    for key, model in models.items():
        key_str = str(key)
        safe_key = re.sub(r"[^A-Za-z0-9._-]", "_", key_str)
        out_path = Path(f"{prefix}{safe_key}.pkl")
        with out_path.open("wb") as out_f:
            pickle.dump(model, out_f, protocol=pickle.HIGHEST_PROTOCOL)
        saved.append(str(out_path))
    
    print(f"Saved {len(saved)} models from {pkl_filename}: {saved}")
```
