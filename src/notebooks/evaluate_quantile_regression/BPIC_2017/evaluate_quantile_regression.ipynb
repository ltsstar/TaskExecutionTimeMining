import time
from line_profiler import LineProfiler
import builtins

def profile_clean_final(duration=60):
    lp = LineProfiler()
    
    # 1. Patch heavy_computation to capture qrm.sample
    original = heavy_computation
    captured = [None]  # mutable for nonlocal
    
    def patched():
        evaluator = conduct_evaluation.ConductEvaluation(
            quantile_regression_models['A'],
            SampleOutcomes_QuantileRegression_A,
            {'case_id_key': 'case:concept:name',
             'activity_key': 'concept:name',
             'resource_key': 'org:resource_start'},
            test_data, n_processes=n_processes, batch_size=batch_size, n=N)
        captured[0] = evaluator.sample_model.qrm.sample
        return evaluator.sample_cases(False, False)
    
    # 2. Add known functions
    from conduct_evaluation import ConductEvaluation
    lp.add_function(ConductEvaluation.sample_case_static)
    
    dummy = SampleOutcomes_QuantileRegression_A(quantile_regression_models['A'], max_sample=10)
    lp.add_function(dummy.sample_from_model)
    
    
    # 4. Add predict() and everything else
    lp.add_function(dummy.qrm.predict)
    for attr in dir(dummy.qrm):
        if callable(getattr(dummy.qrm, attr)) and not attr.startswith('_'):
            try: lp.add_function(getattr(dummy.qrm, attr))
            except: pass
    
    # 5. Replace heavy_computation and profile
    builtins.heavy_computation = patched
    lp.add_function(patched)
    
    lp.enable_by_count()
    start = time.time()
    try:
        while time.time() - start < duration:
            patched()
    except KeyboardInterrupt:
        pass  # silently ignore
    finally:
        lp.disable_by_count()
        builtins.heavy_computation = original  # restore
    
    elapsed = time.time() - start
    print(f"\n{'='*90}")
    print(f"CLEAN PROFILE — {elapsed:.1f}s total runtime")
    print(f"{'='*90}")
    lp.print_stats(stripzeros=True)
    
    print(f"\n{'-'*80}")
    print("TOP 15 HOT LINES (qrm.sample + joblib + sklearn):")
    hot = []
    for (fn, lineno, func), timings in lp.get_stats().timings.items():
        total = sum(t for _,_,t in timings)
        if total > 500_000 and any(k in str(fn) for k in ['sample','predict','joblib','quantile']):
            hot.append((total, func, fn, lineno))
    hot.sort(reverse=True)
    for t, f, fn, l in hot[:15]:
        print(f"{f} @ {fn}:{l} → {t/1e6:.4f}s")
    
    return "DONE"
