```python
import pm4py
import numpy as np
import sys
sys.path.append('../../../TaskExecutionTimeMining/')
from event_log_transformer import *
import pandas
pandas.set_option('display.max_columns', None)
pandas.set_option('display.max_rows', 200)
```

```python
file_path = '../../../../data/BPI_Challenge_2019.xes'
event_log = pm4py.read_xes(file_path)
event_log['id'] = event_log.index
#event_log['Complete Timestamp'] = pandas.to_datetime(event_log['Complete Timestamp'])
```

```python
variants = pm4py.stats.get_variants(event_log)

# Count the number of variants
num_variants = len(variants)

print(f"Number of process variants: {num_variants}")
```

```python
event_counts = event_log.groupby('case:concept:name').size()

# Compute average and standard deviation
avg_events = np.mean(event_counts)
std_events = np.std(event_counts)

print(f"Average number of events per case: {avg_events:.2f}")
print(f"Standard deviation of events per case: {std_events:.2f}")
```

```python
from pm4py.statistics.traces.generic.log import case_statistics


# Get all case durations (in seconds)
durations = case_statistics.get_all_case_durations(event_log)

# Compute average and standard deviation
avg_duration = np.mean(durations) / 3600 / 24
std_duration = np.std(durations) / 3600 / 24

print(f"Average case duration: {avg_duration} days")
print(f"Standard deviation of case durations: {std_duration} days")
```
