```python
import pm4py
import sys
sys.path.append('../../../TaskExecutionTimeMining/')
from event_log_transformer import *
import pandas
pandas.set_option('display.max_columns', None)
pandas.set_option('display.max_rows', 200)
```

```python
file_path = '../../../../data/helpdesk.csv'
event_log = pandas.read_csv(file_path)
event_log['id'] = event_log.index
event_log['Complete Timestamp'] = pandas.to_datetime(event_log['Complete Timestamp'])
```

```python
event_log
```

```python
start_end_event_log = TransformEventLog.start_end_event_log_next(event_log,
                                 merge_activity_on = ['Case ID'],
                                 timestamp_name = 'Complete Timestamp',
                                 start_name_gen = '_start',
                                 complete_name_gen = '_complete',
                                 unique_column = 'id')
```

```python
# Seconds in day
start_end_event_log = TransformEventLog.seconds_in_day(start_end_event_log, 'Complete Timestamp_start')
start_end_event_log = TransformEventLog.day_of_week(start_end_event_log, 'Complete Timestamp_start')
```

```python
# Resource counts
start_end_event_log = TransformEventLog.value_count_per_case_without_lifecycle(start_end_event_log,
                                                             column_name = 'Activity_start',
                                                            case_name = 'Case ID',
                                                            timestamp_name = 'Complete Timestamp_start',
                                                            unique_id = 'id_start')
start_end_event_log = TransformEventLog.value_count_per_case_without_lifecycle(start_end_event_log,
                                                             column_name = 'Resource_start',
                                                            case_name = 'Case ID',
                                                            timestamp_name = 'Complete Timestamp_start',
                                                            unique_id = 'id_start')
start_end_event_log = start_end_event_log.sort_values(by='Complete Timestamp_start')
```

```python
start_end_event_log
```

```python
train, test = pm4py.ml.split_train_test(start_end_event_log, case_id_key='Case ID')
```

```python
print(train.shape, test.shape)
print(train['Case ID'].unique().shape, test['Case ID'].unique().shape)
```

```python
train.to_csv('../../transformed_event_logs/Helpdesk_train.csv', index=False, date_format='%Y-%m-%d %H:%M:%S.%f')
train.to_pickle('../../transformed_event_logs/Helpdesk_train.pickle')
```

```python
test.to_csv('../../transformed_event_logs/Helpdesk_test.csv', index=False, date_format='%Y-%m-%d %H:%M:%S.%f')
test.to_pickle('../../transformed_event_logs/Helpdesk_test.pickle')
```
