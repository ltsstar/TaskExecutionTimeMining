```python
import pm4py
import sys
from pathlib import Path
import pandas

sys.path.append('../../TaskExecutionTimeMining/')
from event_log_transformer import *
pandas.set_option('display.max_columns', None)
pandas.set_option('display.max_rows', 200)
```

```python
file_path = '../../../data/BPIC2015/'

xes_files = list(Path(file_path).rglob('*.xes'))

event_log = None
for xes_file in xes_files:
    print(f"Processing file: {xes_file}")
    # Read the XES file
    new_log = pm4py.read_xes(str(xes_file))
    
    if event_log is None:
        event_log = new_log
    else:
        event_log = pandas.concat([event_log, new_log], ignore_index=True)
```

```python
event_log.columns
```

```python
event_log = event_log.reset_index()
event_log['case_event_id'] = event_log.groupby('case:concept:name').cumcount()

#event_log = event_log[['index', 'case:concept:name', 'concept:name', 'activityNameEN', 'time:timestamp']]
event_log
```

```python
event_log['lifecycle:transition'].unique()
```

```python
start_end_event_log2 = TransformEventLog.start_end_event_log_next(event_log, merge_activity_on=['case:concept:name'], unique_column='index')
start_end_event_log2 = TransformEventLog.seconds_in_day(start_end_event_log2, 'time:timestamp_start')
start_end_event_log2 = TransformEventLog.day_of_week(start_end_event_log2, 'time:timestamp_start')

start_end_event_log2 = start_end_event_log2.sort_values(by=['case:concept:name', 'time:timestamp_start'])
```
